```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
load_dotenv()
```

```python
class BatsmanState(TypedDict):
    runs: int
    balls: int
    fours: int
    sixes: int
    strike_rate: float
    boundary_percentage: float
    balls_per_boundary: float
    
```

```python
def calcaulte_strike_rate(state: BatsmanState) -> BatsmanState:
    # fetch runs and balls 
    runs = state['runs']
    balls = state['balls']

    # calcaulate strike rate
    strike_rate = runs/balls*100

    return {'strike_rate' : strike_rate}

def calculate_balls_per_boundary(state: BatsmanState) -> BatsmanState:
    # fetch outline and title
    balls = state['balls']
    fours = state['fours']
    sixes = state['sixes']

    # calcaulate balls per boundary
    balls_per_boundary = balls/(fours + sixes)
   
    return {'balls_per_boundary' : balls_per_boundary}

def calculate_boundary_percentage(state: BatsmanState) -> BatsmanState:
    # fetch outline and title
    runs = state['runs']
    fours = state['fours']
    sixes = state['sixes']

    # calcaualte boundary percentage
    boundary_percentage = ((fours * 4 + sixes * 6)/runs) * 100

    return {'boundary_percentage' : boundary_percentage}


def summary(state: BatsmanState) -> BatsmanState:
    # fetch outline and title
    summary = f"""
    Strike Rate: {state['strike_rate']}
    Boundary Percentage: {state['boundary_percentage']}
    Balls per Boundary: {state['balls_per_boundary']}
    """

    return {'summary' : summary}
```

```python
# create graph
graph = StateGraph(BatsmanState)

# add nodes
graph.add_node('calcualteStrikeRate', calcaulte_strike_rate)
graph.add_node('calcualteBallsPerBundary', calculate_balls_per_boundary)
graph.add_node('calcuateBoundaryPercentage', calculate_boundary_percentage)
graph.add_node('summary', summary)




# add edges
graph.add_edge(START, "calcualteStrikeRate")
graph.add_edge(START, "calcualteBallsPerBundary")  
graph.add_edge(START, "calcuateBoundaryPercentage") 

graph.add_edge("calcualteStrikeRate", "summary")
graph.add_edge("calcualteBallsPerBundary", "summary")  
graph.add_edge("calcuateBoundaryPercentage", "summary") 

graph.add_edge("summary", END)    

# compile the graph 
workflow  = graph.compile()

# invoke the graph with an initial state

initial_state: BatsmanState = {
    "runs": 100,
    "balls": 50,
    "fours": 6,
    "sixes": 4,
}

state = workflow.invoke(initial_state)
state
```
