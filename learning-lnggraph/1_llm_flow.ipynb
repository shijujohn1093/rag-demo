```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
load_dotenv()
```

```python
model = ChatOpenAI()
```

```python
class LLMState(TypedDict):
    question: str
    answer: str
```

```python
def llm_qa(state: LLMState) -> LLMState:
    # extract the question from state 
    question = state['question']

    # form a prompt
    prompt = f"Answer the following question \n{question}"

    # ask that question to the LLM
    response = model.invoke(prompt).content

    # update the anser in the state
    state['answer'] = response.strip()

    return state
```

```python
# create graph
graph = StateGraph(LLMState)

# add nodes
graph.add_node("llmQa", llm_qa)


# add edges
graph.add_edge(START, "llmQa")
graph.add_edge("llmQa", END)    

# compile the graph 
workflow  = graph.compile()

# invoke the graph with an initial state
state = {"question": "How far the moon from the earth?"}
state = workflow.invoke(state)
state
```
